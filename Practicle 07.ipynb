```python
import re
import string
import nltk
import contractions
import emoji

from bs4 import BeautifulSoup

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
corpus = """
<html>
Hello! I'm John. Don't forget to visit https://www.google.com.
You can email me at john123@gmail.com 😊.

I have 2 dogs and 5 cats!! They are running, jumped, and playing happily.

This is an example of NLP preprocessing.
</html>
"""
```

```python
def preprocess_text(text):


    text = BeautifulSoup(text, "html.parser").get_text()


    text = text.lower()


    text = contractions.fix(text)


    text = re.sub(r'https?://\S+|www\.\S+', '', text)


    text = re.sub(r'\S+@\S+', '', text)


    text = emoji.replace_emoji(text, replace='')


    text = re.sub(r'\d+', '', text)


    text = text.translate(str.maketrans('', '', string.punctuation))


    text = re.sub(r'[^a-zA-Z\s]', '', text)


    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
clean_text = preprocess_text(corpus)

print("Cleaned Text:\n")
print(clean_text)
```

```python
nltk.download('punkt_tab')
tokens = word_tokenize(clean_text)

print("Tokens:\n")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word not in stop_words]

print("After Stopword Removal:\n")
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print("Stemmed Words:\n")
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("Lemmatized Words:\n")
print(lemmatized_words)
```

```python
corpus = """
<html>
नमस्कार! माझे नाव राहुल आहे.
कृपया https://www.google.com ला भेट द्या.
माझा ईमेल rahul123@gmail.com आहे 😊.

माझ्याकडे २ कुत्रे आणि ५ मांजरी आहेत.
ते धावत आहेत, खेळत आहेत आणि आनंदी आहेत.

हे NLP पूर्वप्रक्रियेचे उदाहरण आहे.
</html>
"""
```

```python
def preprocess_text(text):


    text = BeautifulSoup(text, "html.parser").get_text()


    text = re.sub(r'https?://\S+|www\.\S+', '', text)


    text = re.sub(r'\S+@\S+', '', text)


    text = emoji.replace_emoji(text, replace='')


    text = re.sub(r'[0-9०-९]+', '', text)


    text = re.sub(r'[^\w\s]', '', text)


    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
clean_text = preprocess_text(corpus)

print("Cleaned Text:\n")
print(clean_text)
```

```python
tokens = word_tokenize(clean_text)

print("Tokens:\n")
print(tokens)
```

```python
marathi_stopwords = {
    "आहे", "आहेत", "आणि", "हे", "ही", "त्या", "ते",
    "मी", "माझे", "माझा", "माझी", "एक", "ला", "च्या",
    "वर", "मध्ये", "साठी", "होते", "होता", "असून",
    "तर", "किंवा", "पण", "जे", "जो", "ती", "तो",
    "कृपया", "द्या"
}

filtered_tokens = [word for word in tokens if word not in marathi_stopwords]

print("After Stopword Removal:\n")
print(filtered_tokens)
```

```python
def marathi_stem(word):

    suffixes = [
        "ांमध्ये", "ामध्ये", "ांना", "ाना",
        "ाची", "ाचे", "ाचा",
        "तील", "तील",
        "णे", "ला", "ली", "ले",
        "चा", "ची", "चे",
        "ने", "वर", "मध्ये",
        "चा", "च्या", "तील",
        "त", "तो", "ते", "ती"
    ]

    for suffix in suffixes:
        if word.endswith(suffix):
            return word[:-len(suffix)]

    return word

stemmed_words = [marathi_stem(word) for word in filtered_tokens]

print("Stemmed Words:\n")
print(stemmed_words)
```

```python
lemmatized_words = filtered_tokens

print("Lemmatized Words:\n")
print(lemmatized_words)
```
